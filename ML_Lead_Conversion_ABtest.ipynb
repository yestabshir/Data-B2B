Logistic Regression and Random Forest

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Step 1: Generate Sample Data
np.random.seed(42)
n_samples = 1000
data = pd.DataFrame({
    'feature1': np.random.rand(n_samples),
    'feature2': np.random.rand(n_samples) * 100,
    'feature3': np.random.randint(0, 2, size=n_samples),
    'feature4': np.random.normal(0, 1, n_samples),
    'converted': np.random.choice([0, 1], size=n_samples, p=[0.7, 0.3])  # 30% conversion rate
})

# Step 2: Prepare data for training
X = data[['feature1', 'feature2', 'feature3', 'feature4']]
y = data['converted']

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Step 3: Train Logistic Regression Model
log_reg = LogisticRegression()
log_reg.fit(X_train, y_train)

# Step 4: Train Random Forest Model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Step 5: Predict conversion likelihood on test data
log_reg_preds = log_reg.predict(X_test)
rf_preds = rf_model.predict(X_test)

# Step 6: Simulate A/B Test
# Assign leads randomly to either Model A (Logistic Regression) or Model B (Random Forest)
test_data = X_test.copy()
test_data['actual_conversion'] = y_test.values
test_data['assigned_model'] = np.random.choice(['A', 'B'], size=len(test_data))

# Model A (Logistic Regression) predictions
test_data['log_reg_preds'] = log_reg.predict(X_test)
# Model B (Random Forest) predictions
test_data['rf_preds'] = rf_model.predict(X_test)

# Step 7: Calculate Conversion Rates for Each Model
# Group A (Logistic Regression)
group_a = test_data[test_data['assigned_model'] == 'A']
group_a_conversion_rate = accuracy_score(group_a['actual_conversion'], group_a['log_reg_preds'])

# Group B (Random Forest)
group_b = test_data[test_data['assigned_model'] == 'B']
group_b_conversion_rate = accuracy_score(group_b['actual_conversion'], group_b['rf_preds'])

# Output conversion rates for A/B test
print("A/B Test Results")
print(f"Model A (Logistic Regression) Conversion Rate: {group_a_conversion_rate * 100:.2f}%")
print(f"Model B (Random Forest) Conversion Rate: {group_b_conversion_rate * 100:.2f}%")
```

**Explanation:**

1. Data Generation: I have created sample data (Not using the original data) with four features (feature1, feature2, feature3, and feature4) and a target variable converted indicating lead conversion status.

2. Model Training: Logistic Regression and Random Forest models are trained on the dataset to predict converted.

3. A/B Testing Simulation: Leads are assigned randomly to either Model A (Logistic Regression) or Model B (Random Forest) for evaluation.
Each group’s conversion rate is calculated by comparing predicted conversions to actual values.

4. Results: I have calculated and compared the conversion rates for each model to determine which one performed better.
